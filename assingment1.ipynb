```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)

mean = img.mean()

print("Mean intensity =", mean)

_, binary = cv2.threshold(img, mean, 255, cv2.THRESH_BINARY)

print("Grayscale Image:")
cv2_imshow(img)

print("Binary Image:")
cv2_imshow(binary)
```

```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)

threshold = int(input("Enter threshold value (0-255): "))

_, binary = cv2.threshold(
    img,
    threshold,
    255,
    cv2.THRESH_BINARY
)

print("Grayscale Image:")
cv2_imshow(img)

print("Binary Image:")
cv2_imshow(binary)
```
