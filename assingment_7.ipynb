```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("sampleimage.jpg", 0)
img = img.astype(float)

h, w = img.shape
p = np.pad(img, 1, mode='constant')

result = np.zeros((h, w))

# Laplacian mask
mask = np.array([[0, -1, 0],
                 [-1, 4, -1],
                 [0, -1, 0]])

for i in range(h):
    for j in range(w):
        region = p[i:i+3, j:j+3]
        lap = np.sum(region * mask)
        result[i, j] = img[i, j] + lap

result = np.clip(result, 0, 255).astype(np.uint8)

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(result, cmap='gray')
plt.title("Laplacian Sharpening")
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("sampleimage.jpg", 0).astype(float)

h, w = img.shape
p = np.pad(img, 1, mode='constant')

Gx = np.array([[-1, 0, 1],
               [-2, 0, 2],
               [-1, 0, 1]])

Gy = np.array([[-1, -2, -1],
               [ 0,  0,  0],
               [ 1,  2,  1]])

result = np.zeros((h, w))

for i in range(h):
    for j in range(w):
        region = p[i:i+3, j:j+3]

        gx = np.sum(region * Gx)
        gy = np.sum(region * Gy)

        gradient = np.sqrt(gx**2 + gy**2)
        result[i, j] = img[i, j] + gradient

result = np.clip(result, 0, 255).astype(np.uint8)

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(result, cmap='gray')
plt.title("Sobel Sharpening")
plt.axis('off')

plt.show()
```
