```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("sampleimage.jpg", cv2.IMREAD_GRAYSCALE)

# Check if image was loaded
if img is None:
    print("Error: sampleimage.jpg not found")
else:
    width = int(input("Enter border width in pixels: "))

    color = int(input("Enter border color (0-255): "))

    bordered = cv2.copyMakeBorder(
        img,
        width, width, width, width,
        cv2.BORDER_CONSTANT,
        value=color
    )

    print("Original Image:")
    cv2_imshow(img)

    print("Image with Border:")
    cv2_imshow(bordered)





```

```python
import cv2
from google.colab.patches import cv2_imshow

# Read image as grayscale
img = cv2.imread("sampleimage.jpg", cv2.IMREAD_GRAYSCALE)

if img is None:
    print("Image not found")
else:
    # Convert grayscale image to binary
    _, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

    # Binary image complement
    complement = 255 - binary

    print("Original Binary Image:")
    cv2_imshow(binary)

    print("Complemented Binary Image:")
    cv2_imshow(complement)
```
