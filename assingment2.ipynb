```python
import cv2
from google.colab.patches import cv2_imshow


img = cv2.imread("photo.avif")

B, G, R = cv2.split(img)

gray_mean = (R.astype(float) + G.astype(float) + B.astype(float)) / 3
gray_mean = gray_mean.astype('uint8')

print("Original Image:")
cv2_imshow(img)

print("Grayscale - Mean:")
cv2_imshow(gray_mean)


wr = float(input("Enter weightage of R: "))
wg = float(input("Enter weightage of G: "))
wb = float(input("Enter weightage of B: "))


if wr < 0 or wg < 0 or wb < 0 or wr > 1 or wg > 1 or wb > 1:
    print("Weightages must be between 0 and 1")

elif abs((wr + wg + wb) - 1) > 0.0001:
    print("Sum of weightages must be equal to 1")

else:
    gray_weighted = wr * R + wg * G + wb * B
    gray_weighted = gray_weighted.astype('uint8')

    print("Grayscale - Weighted:")
    cv2_imshow(gray_weighted)
```
