```python
4.1

import cv2, numpy as np
import matplotlib.pyplot as plt
from google.colab import files

f = list(files.upload().keys())[0]
img = cv2.imread(f, 0)

log = np.uint8(255 * np.log(1 + img) / np.log(256))

plt.subplot(121), plt.imshow(img, cmap='gray'), plt.title("Original"), plt.axis('off')
plt.subplot(122), plt.imshow(log, cmap='gray'), plt.title("Log"), plt.axis('off')
plt.show()
```

```python
4.2

import cv2, numpy as np
import matplotlib.pyplot as plt
from google.colab import files

f = list(files.upload().keys())[0]
img = cv2.imread(f, 0)

g = float(input("Enter gamma value: "))

power = np.uint8(255 * ((img / 255) ** g))

plt.subplot(121), plt.imshow(img, cmap='gray'), plt.title("Original"), plt.axis('off')
plt.subplot(122), plt.imshow(power, cmap='gray'), plt.title("Power Law"), plt.axis('off')
plt.show()
```

```python
4.3

import cv2, numpy as np
import matplotlib.pyplot as plt
from google.colab import files

f = list(files.upload().keys())[0]
img = cv2.imread(f, 0)

a = int(input("Enter input range start: "))
b = int(input("Enter input range end: "))
c = int(input("Enter output range start: "))
d = int(input("Enter output range end: "))

out = img.copy()
out[(img >= a) & (img <= b)] = ((img[(img >= a) & (img <= b)] - a) * (d - c) / (b - a) + c)

plt.subplot(121), plt.imshow(img, cmap='gray'), plt.title("Original"), plt.axis('off')
plt.subplot(122), plt.imshow(out, cmap='gray'), plt.title("Output"), plt.axis('off')
plt.show()
```
